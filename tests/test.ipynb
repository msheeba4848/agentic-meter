```python
import os
from agentledger import Ledger
from langchain_openai import ChatOpenAI


researcher = ChatOpenAI(model="gpt-4o-mini")
writer = ChatOpenAI(model="gpt-4o-mini")
critic = ChatOpenAI(model="gpt-4o-mini")

with Ledger(budget="$5.00", name="multi_agent_pipeline") as ledger:
    cb = ledger.as_langchain_callback()

    with ledger.tag("researcher"):
        research = researcher.invoke("Research: what is photosynthesis? Two sentences.",
                                     config={"callbacks": [cb]})
    with ledger.tag("writer"):
        draft = writer.invoke(f"Write a haiku based on: {research.content}",
                              config={"callbacks": [cb]})
    with ledger.tag("critic"):
        review = critic.invoke(f"Critique this haiku in one sentence: {draft.content}",
                               config={"callbacks": [cb]})

print(research.content); print(draft.content); print(review.content)
print()
print(ledger.summary())
```
